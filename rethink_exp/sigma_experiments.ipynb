## 1. Check Experiment Status

Checks whether each experiment's `log.txt` exists (indicating completion) and looks for SLURM log files.

````python
def check_status(experiments):
    """Check which experiments have completed by looking for the [perturb] summary line in log.txt."""
    rows = []
    for label, cfg in experiments.items():
        log_path = os.path.join(
            "saved_records", cfg["data_name"], "perturb", cfg["prefix"], "log.txt"
        )
        status = "not started"
        progress = "—"
        
        if os.path.exists(log_path):
            with open(log_path, "r") as f:
                content = f.read()
                lines = content.strip().split("\n")
            
            # Check for the final summary line: [perturb]  obj  regret  time  time
            final_match = re.search(r'\[perturb\]\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s*$', content, re.MULTILINE)
            opt_match = re.search(r'\[Optimal Obj\]:\s*([\d.\-eE+]+)', content)
            
            if final_match:
                status = "DONE"
                regret = float(final_match.group(2))
                if opt_match:
                    optimal = float(opt_match.group(1))
                    pct = (regret / optimal) * 100 if optimal != 0 else float('nan')
                    progress = f"regret={regret:.2f}, %regret={pct:.2f}%"
                else:
                    progress = f"regret={regret:.2f}"
            else:
                status = "running"
                # Find latest epoch from "[perturb] epoch N:" or "Iter N, val"
                epoch_matches = re.findall(r'\[perturb\] epoch (\d+):', content)
                iter_matches = re.findall(r'Iter (\d+), val', content)
                latest_epoch = max(
                    int(epoch_matches[-1]) if epoch_matches else 0,
                    int(iter_matches[-1]) if iter_matches else 0,
                )
                # Find latest val regret
                regret_matches = re.findall(r'val\s+Objective:.*regret:\s+([\d.]+)', content)
                latest_regret = regret_matches[-1] if regret_matches else "?"
                if latest_epoch > 0:
                    progress = f"epoch {latest_epoch}/300, val_regret={latest_regret}"
                else:
                    progress = f"{len(lines)} log lines"
        
        # Check for SLURM error files
        slurm_pattern = f"logs/slurm/*{cfg['prefix']}*"
        slurm_files = sorted(glob.glob(slurm_pattern))
        slurm_info = os.path.basename(slurm_files[-1]) if slurm_files else "—"
        
        rows.append({
            "Experiment": label,
            "Problem": cfg["data_name"],
            "Sigma": cfg["sigma"],
            "LR": cfg.get("lr", "1e-2"),
            "Status": status,
            "Progress": progress,
            "SLURM Log": slurm_info,
        })
    
    df = pd.DataFrame(rows)
    return df

# --- Wave 1 ---
print("=" * 140)
print("Wave 1: Sigma Tuning (lr=1e-2)")
print("=" * 140)
wave1 = {k: v for k, v in EXPERIMENTS.items() if v.get("lr") == "1e-2"}
status_w1 = check_status(wave1)
print(status_w1.to_string(index=False))
n_done1 = (status_w1["Status"] == "DONE").sum()
print(f"\n{n_done1}/{len(wave1)} done")

# --- Wave 2 ---
print(f"\n{'=' * 140}")
print("Wave 2: LR Sweep")
print("=" * 140)
wave2 = {k: v for k, v in EXPERIMENTS.items() if v.get("lr") != "1e-2"}
status_w2 = check_status(wave2)
print(status_w2.to_string(index=False))
n_done2 = (status_w2["Status"] == "DONE").sum()
print(f"\n{n_done2}/{len(wave2)} done")

# Also check SLURM queue
try:
    result = subprocess.run(["squeue", "--me", "--format=%.18i %.30j %.8T %.10M"],
                           capture_output=True, text=True, timeout=10)
    if result.stdout.strip():
        print("\n--- SLURM Queue ---")
        print(result.stdout)
except:
    print("(Could not check SLURM queue)")
````

## 2. Collect Results

Read the final line of each experiment's `log.txt` to extract objective, eval metric (regret), and timing.

````python
def get_results(data_name, prefix):
    """Search log.txt for the final [perturb] summary line and [Optimal Obj] value.
    
    Returns dict with keys: objective, regret, train_time, test_time, optimal_obj, pct_regret
    or None if the experiment hasn't finished.
    """
    log_path = os.path.join(
        "saved_records", data_name, "perturb", prefix, "log.txt"
    )
    if not os.path.exists(log_path):
        return None
    with open(log_path, "r") as f:
        content = f.read()
    
    # Look for the final summary line: [perturb]  obj  regret  train_time  test_time
    perturb_pat = r'\[perturb\]\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s+([\d.\-eE+]+)\s*$'
    match = re.search(perturb_pat, content, re.MULTILINE)
    if not match:
        return None  # experiment still in progress
    
    objective = float(match.group(1))
    regret = float(match.group(2))
    train_time = float(match.group(3))
    test_time = float(match.group(4))
    
    # Extract [Optimal Obj] value from the summary line above
    opt_pat = r'\[Optimal Obj\]:\s*([\d.\-eE+]+)'
    opt_match = re.search(opt_pat, content)
    optimal_obj = float(opt_match.group(1)) if opt_match else np.nan
    
    # Compute %regret = regret / optimal * 100
    if optimal_obj and not np.isnan(optimal_obj) and optimal_obj != 0:
        pct_regret = (regret / optimal_obj) * 100
    else:
        pct_regret = np.nan
    
    return {
        "objective": objective,
        "regret": regret,
        "train_time": train_time,
        "test_time": test_time,
        "optimal_obj": optimal_obj,
        "pct_regret": pct_regret,
    }


def collect_all():
    """Collect results for all registered experiments."""
    rows = []
    for label, cfg in EXPERIMENTS.items():
        res = get_results(cfg["data_name"], cfg["prefix"])
        row = {
            "experiment": label,
            "data_name": cfg["data_name"],
            "sigma": cfg["sigma"],
            "type": cfg["type"],
            "lr": cfg.get("lr", "1e-2"),
        }
        if res is not None:
            row.update(res)
        else:
            for k in ["objective", "regret", "train_time", "test_time", "optimal_obj", "pct_regret"]:
                row[k] = np.nan
        rows.append(row)
    return pd.DataFrame(rows)


results_df = collect_all()
print("=" * 120)
print("Experiment Results")
print("=" * 120)

# Show results grouped by problem
has_data = results_df.dropna(subset=["regret"])
if len(has_data) > 0:
    display_cols = ["experiment", "data_name", "sigma", "type", "lr", "objective", "optimal_obj", "regret", "pct_regret", "train_time"]
    print(has_data[display_cols].to_string(index=False, float_format="%.4f"))
else:
    print("No results available yet. Experiments may still be running.")

print(f"\n{len(has_data)}/{len(results_df)} experiments have results.")
````

## 3. Pairwise Comparisons

Compare constant vs schedule (wave 1) and lr=1e-3 vs lr=5e-3 (wave 2).

````python
def compare_pairs(results_df, pairs_dict, title="Comparison"):
    """Build a comparison table for any set of (key_a, key_b) pairs."""
    rows = []
    for problem_name, (key_a, key_b) in pairs_dict.items():
        a_row = results_df[results_df["experiment"] == key_a]
        b_row = results_df[results_df["experiment"] == key_b]
        
        a_regret = a_row["regret"].values[0] if len(a_row) else np.nan
        b_regret = b_row["regret"].values[0] if len(b_row) else np.nan
        a_pct    = a_row["pct_regret"].values[0] if len(a_row) else np.nan
        b_pct    = b_row["pct_regret"].values[0] if len(b_row) else np.nan
        a_opt    = a_row["optimal_obj"].values[0] if len(a_row) else np.nan
        b_opt    = b_row["optimal_obj"].values[0] if len(b_row) else np.nan
        optimal  = a_opt if not np.isnan(a_opt) else b_opt
        
        a_cfg = EXPERIMENTS[key_a]
        b_cfg = EXPERIMENTS[key_b]
        a_label = f"{a_cfg['type']} σ={a_cfg['sigma']} lr={a_cfg['lr']}"
        b_label = f"{b_cfg['type']} σ={b_cfg['sigma']} lr={b_cfg['lr']}"
        
        if np.isnan(a_regret) or np.isnan(b_regret):
            winner = "—"
        elif a_regret < b_regret:
            winner = key_a
        elif b_regret < a_regret:
            winner = key_b
        else:
            winner = "tie"
        
        rows.append({
            "Problem": problem_name,
            "Optimal": optimal,
            f"A ({key_a})": a_label,
            "Regret(A)": a_regret,
            "%Regret(A)": a_pct,
            f"B ({key_b})": b_label,
            "Regret(B)": b_regret,
            "%Regret(B)": b_pct,
            "Winner": winner,
        })
    return pd.DataFrame(rows)


# Wave 1 comparison
print("=" * 130)
print("Wave 1: Constant vs Schedule+Tanh  (lower regret = better)")
print("=" * 130)
comp1 = compare_pairs(results_df, PROBLEM_PAIRS, "Wave 1")
print(comp1.to_string(index=False, float_format="%.4f"))

# Wave 2 comparison
print(f"\n{'=' * 130}")
print("Wave 2: LR Sweep  (lower regret = better)")
print("=" * 130)
comp2 = compare_pairs(results_df, LR_PAIRS, "Wave 2")
print(comp2.to_string(index=False, float_format="%.4f"))
````

## 4. Compare with Baseline Methods

Load existing baseline results (MSE, DFL, Blackbox, etc.) and see where our perturb experiments rank.

````python
# Baseline method configs (from collect_with_perturb.py)
BASELINE_CONFIG = {
    "budgetalloc-real":       ["mse", "dfl", "blackbox", "identity", "spo", "nce", "pointLTR", "listLTR", "pairLTR", "lodl"],
    "bipartitematching-cora": ["bce", "dfl", "blackbox", "identity", "cpLayer", "spo", "nce", "pointLTR", "listLTR", "pairLTR", "lodl"],
    "knapsack-gen":           ["mse", "dfl", "blackbox", "identity", "cpLayer", "spo", "nce", "pointLTR", "listLTR", "pairLTR", "lodl"],
    "knapsack-energy":        ["mse", "dfl", "blackbox", "identity", "cpLayer", "spo", "nce", "pointLTR", "listLTR", "pairLTR", "lodl"],
    "portfolio-real":         ["mse", "dfl", "blackbox", "identity", "cpLayer", "spo", "nce", "pointLTR", "listLTR", "pairLTR", "lodl"],
    "energy-energy":          ["mse", "dfl", "blackbox", "identity", "spo", "nce", "pointLTR", "pairLTR", "listLTR", "lodl"],
}

def get_baseline_results(data_name, method, prefix="bench"):
    """Read baseline results from saved_records/<data_name>/<method>/<prefix>/log.txt.
    Returns dict with objective, regret, optimal_obj, pct_regret or None."""
    log_path = os.path.join("saved_records", data_name, method, prefix, "log.txt")
    if not os.path.exists(log_path):
        return None
    with open(log_path, "r") as f:
        content = f.read()
    
    # Try to find the method's final summary line: [<method>]  obj  regret  train_time  test_time
    lines = content.strip().split("\n")
    last_line = lines[-1].strip() if lines else ""
    parts = last_line.split("  ")
    parts = [p.strip() for p in parts if p.strip()]
    try:
        vals = [float(x) for x in parts[-4:]]
    except:
        return None
    
    # Extract [Optimal Obj]
    opt_match = re.search(r'\[Optimal Obj\]:\s*([\d.\-eE+]+)', content)
    optimal_obj = float(opt_match.group(1)) if opt_match else np.nan
    pct_regret = (vals[1] / optimal_obj) * 100 if (optimal_obj and not np.isnan(optimal_obj) and optimal_obj != 0) else np.nan
    
    return {"objective": vals[0], "regret": vals[1], "optimal_obj": optimal_obj, "pct_regret": pct_regret}


# Build combined table per problem (unique data_names)
seen_data_names = []
for cfg in EXPERIMENTS.values():
    if cfg["data_name"] not in seen_data_names:
        seen_data_names.append(cfg["data_name"])

for data_name in seen_data_names:
    # Gather all our experiments for this data_name
    our_keys = [k for k, v in EXPERIMENTS.items() if v["data_name"] == data_name]
    
    print(f"\n{'=' * 100}")
    print(f"  {data_name}")
    print(f"{'=' * 100}")
    
    rows = []
    optimal_obj = np.nan
    
    # Baselines
    if data_name in BASELINE_CONFIG:
        for method in BASELINE_CONFIG[data_name]:
            res = get_baseline_results(data_name, method)
            if res is not None:
                rows.append({"method": method, "objective": res["objective"], "regret": res["regret"],
                             "%regret": res["pct_regret"], "source": "baseline"})
                if not np.isnan(res["optimal_obj"]):
                    optimal_obj = res["optimal_obj"]
    
    # Our perturb experiments for this problem
    for key in our_keys:
        ecfg = EXPERIMENTS[key]
        res = get_results(ecfg["data_name"], ecfg["prefix"])
        label = f"perturb ({ecfg['type']}, σ={ecfg['sigma']}, lr={ecfg['lr']})"
        if res is not None:
            rows.append({"method": label, "objective": res["objective"], "regret": res["regret"],
                         "%regret": res["pct_regret"], "source": "★ ours"})
            if not np.isnan(res["optimal_obj"]):
                optimal_obj = res["optimal_obj"]
        else:
            rows.append({"method": label, "objective": np.nan, "regret": np.nan,
                         "%regret": np.nan, "source": "★ ours"})
    
    if rows:
        tbl = pd.DataFrame(rows).sort_values("regret", na_position="last")
        tbl.insert(0, "rank", range(1, len(tbl) + 1))
        opt_str = f"  [Optimal Obj = {optimal_obj:.4f}]" if not np.isnan(optimal_obj) else ""
        print(f"{opt_str}")
        print(tbl[["rank", "method", "regret", "%regret", "objective", "source"]].to_string(index=False, float_format="%.4f"))
    else:
        print("  (no results available)")
````

## 5. Visualizations

````python
def plot_lr_comparison(results_df):
    """For each problem, bar-chart comparing lr=1e-2, 1e-3, 5e-3 for the best config type."""
    # Group experiments by data_name
    problems = []
    for dn in dict.fromkeys(cfg["data_name"] for cfg in EXPERIMENTS.values()):
        exps = {k: v for k, v in EXPERIMENTS.items() if v["data_name"] == dn}
        # Get results for each
        lr_results = {}
        for key, cfg in exps.items():
            res = get_results(cfg["data_name"], cfg["prefix"])
            if res is not None:
                lr_results[key] = {**cfg, **res}
        if lr_results:
            problems.append((dn, lr_results))
    
    if not problems:
        print("No results to plot yet.")
        return
    
    n_probs = len(problems)
    fig, axes = plt.subplots(1, n_probs, figsize=(max(4 * n_probs, 12), 5), squeeze=False)
    axes = axes[0]
    
    LR_COLORS = {"1e-2": "#4C72B0", "1e-3": "#55A868", "5e-3": "#DD8452"}
    
    for idx, (dn, lr_results) in enumerate(problems):
        ax = axes[idx]
        # Group by (type, sigma)
        groups = {}
        for key, info in lr_results.items():
            group_key = f"{info['type']}\nσ={info['sigma']}"
            if group_key not in groups:
                groups[group_key] = {}
            groups[group_key][info.get("lr", "1e-2")] = info["pct_regret"]
        
        x = np.arange(len(groups))
        width = 0.25
        all_lrs = sorted(set(lr for g in groups.values() for lr in g.keys()))
        
        for i, lr in enumerate(all_lrs):
            vals = [groups[g].get(lr, np.nan) for g in groups]
            offset = (i - len(all_lrs)/2 + 0.5) * width
            bars = ax.bar(x + offset, vals, width, label=f"lr={lr}",
                         color=LR_COLORS.get(lr, f"C{i}"), alpha=0.85)
            for bar in bars:
                h = bar.get_height()
                if not np.isnan(h):
                    ax.annotate(f"{h:.1f}%", (bar.get_x() + bar.get_width()/2, h),
                               ha="center", va="bottom", fontsize=7)
        
        ax.set_title(dn, fontsize=10)
        ax.set_xticks(x)
        ax.set_xticklabels(list(groups.keys()), fontsize=7, rotation=30, ha="right")
        ax.set_ylabel("%Regret")
        ax.legend(fontsize=7)
        ax.grid(axis="y", alpha=0.3)
    
    plt.suptitle("Learning Rate Comparison: %Regret by Problem", fontsize=13)
    plt.tight_layout()
    plt.savefig("saved_records/lr_comparison.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: saved_records/lr_comparison.png")


plot_lr_comparison(results_df)
````

````python
def plot_ranking_with_baselines(results_df):
    """Heatmap showing where perturb experiments rank among all baselines, per data_name."""
    all_problems = []
    all_methods = set()
    problem_results = {}
    
    seen_data_names = list(dict.fromkeys(cfg["data_name"] for cfg in EXPERIMENTS.values()))
    
    for data_name in seen_data_names:
        method_regrets = {}
        
        # Baselines
        if data_name in BASELINE_CONFIG:
            for method in BASELINE_CONFIG[data_name]:
                res = get_baseline_results(data_name, method)
                if res is not None:
                    method_regrets[method] = res["regret"]
                    all_methods.add(method)
        
        # Our experiments for this data_name
        for key, cfg in EXPERIMENTS.items():
            if cfg["data_name"] != data_name:
                continue
            res = get_results(cfg["data_name"], cfg["prefix"])
            if res is not None:
                label = f"perturb_{key}"
                method_regrets[label] = res["regret"]
                all_methods.add(label)
        
        if method_regrets:
            problem_results[data_name] = method_regrets
            all_problems.append(data_name)
    
    if not problem_results:
        print("No results to plot yet.")
        return
    
    # Build regret matrix
    all_methods_sorted = sorted(all_methods)
    matrix = pd.DataFrame(index=all_methods_sorted, columns=all_problems, dtype=float)
    for prob, mdict in problem_results.items():
        for m, val in mdict.items():
            matrix.loc[m, prob] = val
    
    # Compute ranks per problem
    rank_matrix = matrix.rank(axis=0, method="min", na_option="bottom")
    rank_matrix["avg_rank"] = rank_matrix.mean(axis=1)
    rank_matrix = rank_matrix.sort_values("avg_rank")
    
    print("\nRankings (lower = better):")
    print(rank_matrix.to_string())
    
    # Heatmap
    plot_data = rank_matrix.drop(columns=["avg_rank"])
    fig, ax = plt.subplots(figsize=(max(8, len(all_problems) * 1.5), max(6, len(all_methods_sorted) * 0.35)))
    
    im = ax.imshow(plot_data.values, cmap="RdYlGn_r", aspect="auto")
    ax.set_xticks(range(len(plot_data.columns)))
    ax.set_xticklabels(plot_data.columns, rotation=30, ha="right")
    ax.set_yticks(range(len(plot_data.index)))
    ax.set_yticklabels(plot_data.index, fontsize=7)
    
    # Annotate cells with rank values
    for i in range(len(plot_data.index)):
        for j in range(len(plot_data.columns)):
            val = plot_data.iloc[i, j]
            if not np.isnan(val):
                method_name = plot_data.index[i]
                weight = "bold" if "perturb" in method_name else "normal"
                ax.text(j, i, f"{int(val)}", ha="center", va="center",
                        fontsize=8, fontweight=weight)
    
    plt.colorbar(im, ax=ax, label="Rank")
    ax.set_title("Method Rankings by Problem (lower rank = better)")
    plt.tight_layout()
    plt.savefig("saved_records/sigma_experiment_rankings.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved: saved_records/sigma_experiment_rankings.png")


plot_ranking_with_baselines(results_df)
````

## 6. Training Curve Analysis

Parse `saved_records/<data_name>/perturb/<prefix>/log.txt` for per-epoch regret and sigma values.
Uses data_name-specific paths to avoid cross-contamination between problems that share the same prefix (e.g. knapsack-gen vs knapsack-energy).

````python
def parse_training_log(data_name, prefix):
    """Parse saved_records/<data_name>/perturb/<prefix>/log.txt for training curves."""
    log_path = os.path.join("saved_records", data_name, "perturb", prefix, "log.txt")
    if not os.path.exists(log_path):
        return None
    
    epochs = []
    with open(log_path, "r") as f:
        current_epoch = {}
        for line in f:
            m = re.search(r'\[perturb\] epoch (\d+): \|coeff_hat\| = ([\d.]+), sigma = ([\d.]+)', line)
            if m:
                current_epoch = {
                    "epoch": int(m.group(1)),
                    "coeff_hat": float(m.group(2)),
                    "sigma": float(m.group(3)),
                }
            
            # Parse train line: Iter N, train Objective: X, Loss: Y Pred Loss: Z, regret: W
            m_train = re.search(r'Iter \d+, train\s+Objective:\s+(-?[\d.]+).*regret:\s+([\d.]+)', line)
            if m_train and current_epoch:
                current_epoch["train_objective"] = float(m_train.group(1))
                current_epoch["train_regret"] = float(m_train.group(2))
            
            # Parse val line: Iter N, val Objective: X, ... regret: W
            m_val = re.search(r'Iter \d+, val\s+Objective:\s+(-?[\d.]+).*regret:\s+([\d.]+)', line)
            if m_val and current_epoch:
                current_epoch["val_objective"] = float(m_val.group(1))
                current_epoch["val_regret"] = float(m_val.group(2))
                epochs.append(current_epoch)
                current_epoch = {}
    
    if not epochs:
        return None
    return pd.DataFrame(epochs)


# Plot training curves for ALL_PAIRS - now with 3 rows: train obj, train regret, val regret
n_pairs = len(ALL_PAIRS)
n_cols = 4
n_rows_per_metric = (n_pairs + n_cols - 1) // n_cols
n_metrics = 3  # train_objective, train_regret, val_regret

fig, axes = plt.subplots(n_metrics * n_rows_per_metric, n_cols, figsize=(20, 4 * n_metrics * n_rows_per_metric))

COLORS = ["#4C72B0", "#DD8452", "#55A868", "#C44E52", "#8172B3", "#937860"]
METRICS = [
    ("train_objective", "Train Objective"),
    ("train_regret", "Train Regret"),
    ("val_regret", "Val Regret"),
]

for metric_idx, (metric_key, metric_label) in enumerate(METRICS):
    for idx, (problem_name, (key_a, key_b)) in enumerate(ALL_PAIRS.items()):
        row_offset = metric_idx * n_rows_per_metric
        ax = axes[row_offset + idx // n_cols, idx % n_cols]
        has_data = False
        
        for i, key in enumerate([key_a, key_b]):
            cfg = EXPERIMENTS[key]
            label = f"{cfg['type']} σ={cfg['sigma']} lr={cfg['lr']}"
            curve = parse_training_log(cfg["data_name"], cfg["prefix"])
            if curve is not None and len(curve) > 0 and metric_key in curve.columns:
                ax.plot(curve["epoch"], curve[metric_key], label=label, color=COLORS[i % len(COLORS)], alpha=0.8)
                has_data = True
        
        ax.set_title(problem_name, fontsize=9)
        ax.set_xlabel("Epoch")
        ax.set_ylabel(metric_label)
        if has_data:
            ax.legend(fontsize=6)
        else:
            ax.text(0.5, 0.5, "No data yet", ha="center", va="center", transform=ax.transAxes, fontsize=11, color="gray")
        ax.grid(alpha=0.3)

# Hide unused axes in each metric section
for metric_idx in range(n_metrics):
    row_offset = metric_idx * n_rows_per_metric
    for col_idx in range(len(ALL_PAIRS) % n_cols, n_cols):
        if len(ALL_PAIRS) % n_cols != 0:
            axes[row_offset + n_rows_per_metric - 1, col_idx].set_visible(False)

# Add row labels on the left
for metric_idx, (metric_key, metric_label) in enumerate(METRICS):
    row_offset = metric_idx * n_rows_per_metric
    fig.text(0.01, 1 - (metric_idx + 0.5) / n_metrics, metric_label, fontsize=12, fontweight='bold',
             va='center', ha='left', rotation=90)

plt.suptitle("Training Curves: Objective and Regret per Epoch", fontsize=14, y=1.01)
plt.tight_layout()
plt.savefig("saved_records/sigma_experiment_curves.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: saved_records/sigma_experiment_curves.png")
````
